# Inbyggda _magic commands_

- Rad-magic (`%`): verkar på en enskild rad
- Cell-magic (`%%`): verkar på en hel cell
- Mer information: https://ipython.readthedocs.io/en/stable/interactive/magics.html

Detta kan t.ex. användas för att ta tiden för att köra en cell. Här vill vi beräkna summan
$\sum_{n=1}^{k} \frac{1}{n^2}$. Termerna $1/n^2$ dyker bl.a. upp i energinivåerna för
[Bohrs atommodell](https://sv.wikipedia.org/wiki/Bohrs_atommodell), där energin är proportionell mot $1/n^2$.

```python
def minfunktion(k):          # detta är en funktion (function)
    x = 0.0
    for n in range(1, k + 1):  # n går från 1 till k
        x = x + 1.0 / n**2
    return x

print('f =', minfunktion(100))
```

```python
%time minfunktion(100)
```

`%time` kör koden **en** gång och rapporterar den tiden. `%timeit` kör koden **många** gånger
och rapporterar ett genomsnitt (samt det bästa resultatet), vilket ger ett mer tillförlitligt mått
för snabb kod. Jämför de två cellerna nedan.

```python
%%timeit
minfunktion(100)
```
