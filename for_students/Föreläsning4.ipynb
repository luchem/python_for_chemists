```python
import random
```

Nu kan du använda funktioner och submoduler i `random` med `random.funktion`, till exempel `random.gauss` som ger ett slumptal enligt en gaussisk fördelning.

```python
help(random.gauss)

mu = 5.0
sigma = 0.1
print(random.gauss(mu, sigma))
```

Men du kan också importera bara det du behöver:

```python
from random import gauss
print(gauss(mu, sigma)) # Nu behöver du inte random.gauss men bara gauss
```

Du kan också ge ett eget namn åt funktionen/biblioteket du importerar:

```python
from random import gauss as randgauss

print(randgauss(mu, sigma))
```

Vissa bibliotek kommer direkt med Python-installationen, medan andra behöver laddas ner separat:

* **Standardbibliotek:** Python kommer med ett omfattande standardbibliotek som innehåller många moduler och verktyg för att utföra olika uppgifter. Dessa bibliotek är utvecklade och underhållna av Python-gemenskapen och är tillgängliga direkt när du installerar Python. Exempel på sådana bibliotek är `math`, `os`, `datetime`, `random` osv. Du kan använda dessa bibliotek utan att behöva installera något ytterligare.
* **Tredjepartsbibliotek:** Utöver standardbiblioteket finns det tusentals tredjepartsbibliotek som utvecklats av olika organisationer eller enskilda utvecklare runt om i världen. Dessa bibliotek erbjuder mer specialiserade funktioner och verktyg för olika ändamål, såsom datavisualisering, maskininlärning, webbutveckling osv. För att använda dem måste du ladda ner och installera dem separat. Exempel på sådana bibliotek är `NumPy`, `Pandas`, `Matplotlib`, `RDKit` och många fler.

För att installera tredjepartsbibliotek kan du använda pakethanterarna `pip` eller `conda`.

## Rita grafer (plotta)
En viktig del av vetenskap är att på ett bra och vettigt sätt kunna visualisera sin data. I Python görs detta enkelt med ett bibliotek som heter `matplotlib`. Vi ska nu testa och se vilka funktionaliteter detta bibliotek har. Vi börjar med att importera det och samtidigt skapa kortkommandot `plt` (för plot).

```python
import matplotlib
import matplotlib.pyplot as plt
```

### Vi börjar genom att skapa två listor med tal

```python
C = 1.
k=0.2
dt = 0.01
T=1000
Cr = []
t = []
Cr.append(C)
t.append(0)
for i in range(0,T):
    C = C*(1.-k*dt)
    Cr.append(C+random.gauss(0.0,0.05)) # Här simulerar vi experimentellt "brus"
    t.append((i+1)*dt)
```

## plot()
Vi använder nu `plt.plot()` för att visualisera vår data. Som ni ser nedan tar `plt.plot()` i detta fall två argument: det första är $x$-värdena och det andra $y$-värdena. Funktionen `plt.show()` visar grafen.
I Jupyter kan bilden enkelt sparas genom att dra figuren till en katalog eller skrivbordet.

```python
plt.plot(t,Cr)
plt.show()
```

### Man kan enkelt ändra stilen på sina linjer, eller välja att visa dem som symboler

I detta fall ändrar vi `linestyle`, som också förkortas `ls`.

```python
plt.plot(t,Cr,ls='-')
plt.show()
```

# Histogram, scatter-plottar, felstaplar och inset

Två andra typer av grafer är histogram och scatter-plottar.

# Felstaplar

I nedanstående fall sparar vi nu även våra fel. I detta fall har vi genererat felen (enligt en normalfördelning). Men i många fall har vi en uppfattning om felet genom att göra upprepade experiment.

```python
C = 1.
k=0.2
dt = 0.01
T=1000
Cr = []
Er = []
t = []
tr = []
Err = []
Cr.append(C)
Er.append(0)
Err.append(0)
t.append(0)
tr.append(0)
for i in range(0,T):
    C = C*(1.-k*dt)
    E = random.gauss(0.0,0.05) # Här simulerar vi experimentellt "brus"
    Et = random.gauss(0.0,0.02) # Här simulerar vi experimentellt "brus"
    Er.append(abs(2*E))
    Err.append(E)
    Cr.append(C+E) 
    t.append((i+1)*dt+Et)
    tr.append(Et)
```

### Felet i koncentrationen
Vi börjar med att titta på felet vi lagt på (eller uppmätt) i vår koncentration. I detta fall ritar vi grafen med hjälp av `scatter()`-funktionen, som markerar ut våra punkter.

```python
plt.scatter(t,Err)
plt.show()
```

### Vi kan även anta att vi har en mätosäkerhet i tidtagningen.

Låt oss nu generera en graf som visar felen för varje mätpunkt.
På $x$-axeln har vi en punkts fel i tid och på $y$-axeln felet i koncentration.
Vi ser att vi får en massa punkter med en högre täthet i mitten.

```python
plt.scatter(tr,Err)
plt.show()
```

# 2D-grafer med matplotlib och numpy
Vi kommer att lära oss mer om `NumPy` nästa föreläsning. Detta bibliotek är väldigt användbart.
I detta fall använder vi numpy för att läsa in en matris med tal. I det specifika fallet är vi intresserade av sannolikheten att hitta en jon i närheten av en annan på en tvådimensionell yta.
Vi gör detta med hjälp av `plt.imshow()`. Figuren liknar rätt mycket 2D-stapeldiagrammet vi gjorde tidigare. I detta fall är det en så kallad korrelationsfunktion mellan laddade partiklar som fastnat på en motsatt laddad yta.

```python
import numpy as np
Gr2d = np.loadtxt("gr2dwigner.dat")
print(Gr2d)

im = plt.imshow(Gr2d/Gr2d.mean(), cmap="RdBu", vmin=max(2.-Gr2d.max()/Gr2d.mean(),0), vmax=Gr2d.max()/Gr2d.mean(),extent=[-600, 600, -600, 600])
plt.colorbar()
plt.show()
```

# 3D-grafer
Det är även möjligt att göra 3D-grafer (även om det oftast inte är att rekommendera, då de ofta är svårtolkade).

```python
from mpl_toolkits.mplot3d import Axes3D
fig= plt.figure(figsize=(16,8))
ax= fig.add_subplot(projection='3d')
#print(Gr2d.shape)
x = np.linspace(-600,600,139,endpoint=True)
y = np.linspace(-600,600,139,endpoint=True)
X,Y=np.meshgrid(x,y)
Z = Gr2d/Gr2d.mean()

surf=ax.plot_surface(X,Y,Z,cmap='jet',antialiased=False)
#ax.contourf(X, Y, Z, zdir='z',offset=0., cmap='jet')
#ax.contourf(X, Y, Z, zdir='x',offset=-600,cmap='jet')
#ax.contourf(X, Y, Z, zdir='y',offset=600, cmap='jet')
fig.colorbar(surf, shrink=0.5, aspect=10)
ax.set_zlim3d(-0,2)
ax.set_xlabel('x-axis',labelpad=15)
ax.set_ylabel('y-axis',labelpad=15)
ax.set_zlabel('gr-axis',labelpad=15)
plt.show()
```
